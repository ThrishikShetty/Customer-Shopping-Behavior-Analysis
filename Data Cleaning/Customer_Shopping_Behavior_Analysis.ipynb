```python
# Checking if missing data or null values are present in the dataset

df.isnull().sum()
```

### Analysis

- `df.isnull().sum()` is used to check the number of missing values in each column.
- The dataset contains **37 missing values in the `Review Rating` column**.
- All other columns have **0 missing values**, indicating that they are complete.
- Since the missing values are limited to one column, we can handle them separately without affecting the rest of the dataset.
- Before performing analysis involving `Review Rating`, the missing values should be appropriately handled.

```python
# Imputing missing values in Review Rating column with the median rating of the product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

### Analysis

- The missing values in the `Review Rating` column are being handled using **median imputation**.
- The median rating is calculated **separately for each product category** using `groupby('Category')`.
- Missing ratings are then replaced with the **median rating of their respective category**.
- This approach is better than using the overall median because customers in different categories may have different typical rating patterns.
- `transform()` ensures that the category-level median is aligned with each original row while keeping the **same number of rows** in the DataFrame.
- Using the median also helps reduce the influence of unusually high or low ratings.

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

### Analysis

- The column names are standardized to **lowercase** to make them easier to read and work with.
- Spaces in column names are replaced with **underscores (`_`)**, following the **snake_case naming convention**.
- The column `purchase_amount_(usd)` is renamed to **`purchase_amount`** for simplicity.
- Standardized column names make the dataset easier to use in **Python, SQL, and data visualization**.
- This also improves code readability and avoids having to use spaces or special characters when referencing columns.

```python
bins = [0, 25, 35, 50, 100]
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.cut(
    df['age'],
    bins=bins,
    labels=labels
)
```

### Analysis

- The `age` column is divided into **four age groups** using `pd.cut()`.
- The defined age ranges are:
  - **0–25:** Young Adult
  - **25–35:** Adult
  - **35–50:** Middle-aged
  - **50–100:** Senior
- A new column called **`age_group`** is created based on these age ranges.
- Converting individual ages into groups makes it easier to analyze and compare **purchasing behavior across different age segments**.
- These age groups can later be used for **groupby analysis, visualizations, and customer segmentation**.

```python
df[['age','age_group']].head(10)
```

### Analysis

- The output shows the original `age` values along with their corresponding `age_group`.
- This confirms that each customer's age has been successfully categorized into the appropriate age segment.
- The `age_group` column can now be used to compare **customer behavior, purchase amounts, and preferences across different age groups**.
- Grouping customers by age makes the dataset easier to analyze and can help identify **age-based purchasing patterns**.

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

### Analysis

- A new column, **`purchase_frequency_days`**, is created to convert purchase frequency categories into numerical values.
- Each frequency is mapped to the approximate number of days between purchases.
- For example, **Weekly = 7 days**, **Monthly = 30 days**, **Quarterly = 90 days**, and **Annually = 365 days**.
- `map()` is used to replace each categorical frequency with its corresponding number of days.
- Converting purchase frequency into a numerical format makes it easier to perform **calculations, comparisons, customer segmentation, and frequency-based analysis**.

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

### Analysis

- The output confirms that the `frequency_of_purchases` categories have been successfully converted into numerical values in `purchase_frequency_days`.
- For example, **Fortnightly = 14 days**, **Weekly = 7 days**, **Quarterly = 90 days**, and **Annually = 365 days**.
- This confirms that the mapping has been applied correctly to the displayed records.
- The new numerical column can now be used for **calculations, comparisons, and customer purchasing frequency analysis**.

```python
from sqlalchemy import create_engine

# Step 1: Connect to PostgreSQL
# Replace placeholders with your actual details
username = "postgres"      # default user
password = "root123" # the password you set during installation
host = "localhost"         # if running locally
port = "5433"              # default PostgreSQL port
database = "customer_behavior"    # the database you created in pgAdmin

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# Step 2: Load DataFrame into PostgreSQL
table_name = "customer"   # choose any table name
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```

### Analysis

- `create_engine()` is used to establish a connection between Python and the **PostgreSQL database** using SQLAlchemy.
- The database connection is configured with the **username, password, host, port, and database name**.
- `df.to_sql()` is used to transfer the cleaned DataFrame from Python into PostgreSQL.
- The DataFrame is stored as a table named **`customer`** in the `customer_behavior` database.
- `if_exists='replace'` means that if a table named `customer` already exists, it will be **replaced with the current DataFrame**.
- `index=False` prevents the Pandas DataFrame index from being added as an unnecessary column in the PostgreSQL table.
- The printed message confirms that the data has been successfully loaded into PostgreSQL.
- This allows the cleaned dataset to be further analyzed using **SQL queries**.
